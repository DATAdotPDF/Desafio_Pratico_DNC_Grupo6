```python
# 📄 Listar as abas do Excel
abas = pd.ExcelFile(caminho_arquivo).sheet_names

# 🏷️ Colunas categóricas e numéricas
coluna_categoria = 'Nível'
colunas_numericas = [
    'Total',
    'Algodão herbáceo (em caroço)',
    'Arroz (em casca)',
    'Cacau (em amêndoa)',
    'Café (em grão) Total',
    'Cana-de-açúcar',
    'Feijão (em grão)',
    'Fumo (em folha)',
    'Laranja',
    'Milho (em grão)',
    'Soja (em grão)'
]

# 📦 Dicionário para armazenar os DataFrames
dfs = {}

# 🔁 Loop pelas abas do Excel
for nome_aba in abas:
    # Nome do DataFrame com ano
    nome_df = f"df_2022_{nome_aba.lower().replace(' ', '_').replace('-', '_').replace('(', '').replace(')', '')}"

    # 📥 Ler aba
    df_temp = pd.read_excel(caminho_arquivo, sheet_name=nome_aba)

    # 🏷️ Tipagem da coluna categórica
    if coluna_categoria in df_temp.columns:
        df_temp[coluna_categoria] = df_temp[coluna_categoria].astype('category')

    # 🔢 Tipagem de colunas numéricas
    for col in colunas_numericas:
        if col in df_temp.columns:
            df_temp[col] = pd.to_numeric(df_temp[col], errors='coerce')

    # 📆 Adicionar coluna de ano
    df_temp['Ano'] = 2022

    # 💾 Salvar no dicionário
    dfs[nome_df] = df_temp
```

```python
# 📦 Dicionário final para armazenar os DataFrames por aba e nível
dfs_nivel = {}

# 🔁 Iterar sobre os DataFrames já carregados e tipados
for nome_df, df in dfs.items():

    # ✅ Garantir que a coluna 'Nível' exista e seja categórica
    if 'Nível' in df.columns:
        df['Nível'] = df['Nível'].astype('category')

        # 🕒 Extrair o ano a partir do nome do DataFrame
        if "2022" in nome_df:
            ano = 2022
        elif "2023" in nome_df:
            ano = 2023
        else:
            ano = None  # fallback para debugging se necessário

        # 🏷️ Adicionar coluna 'Ano' ao DataFrame
        df['Ano'] = ano

        # 🔢 Colunas numéricas a serem convertidas
        colunas_numericas = [
            'Total',
            'Algodão herbáceo (em caroço)',
            'Arroz (em casca)',
            'Cacau (em amêndoa)',
            'Café (em grão) Total',
            'Cana-de-açúcar',
            'Feijão (em grão)',
            'Fumo (em folha)',
            'Laranja',
            'Milho (em grão)',
            'Soja (em grão)'
        ]

        for col in colunas_numericas:
            if col in df.columns:
                df[col] = pd.to_numeric(df[col], errors='coerce')

        # 🔍 Separar e salvar por cada nível único encontrado
        for nivel in df['Nível'].dropna().unique():
            nivel_nome = str(nivel).strip().lower().replace(" ", "_")
            nome_final = f"{nome_df}_{nivel_nome}"

            # Criar cópia do DataFrame filtrado
            df_nivel = df[df['Nível'] == nivel].copy()

            # Armazenar no dicionário
            dfs_nivel[nome_final] = df_nivel

# ✅ Visualizar os primeiros nomes gerados
for nome in list(dfs_nivel.keys())[:10]:
    print(nome)
```

```python
import pandas as pd
import os

# Caminhos
caminho_arquivo_2023 = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/RAW/2023/(2023) -  SIDRA Top10 - 5457.xlsx'
caminho_saida = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/2023'

# Parâmetros
coluna_categoria = 'Nível'
colunas_numericas = [
    'Algodão herbáceo (em caroço)',
    'Arroz (em casca)',
    'Cacau (em amêndoa)',
    'Café (em grão) Total',
    'Cana-de-açúcar',
    'Feijão (em grão)',
    'Fumo (em folha)',
    'Laranja',
    'Milho (em grão)',
    'Soja (em grão)'
]
ano_referencia = 2023

# Listar abas
abas_2023 = pd.ExcelFile(caminho_arquivo_2023).sheet_names
dfs_2023 = {}

# Carregar e tratar cada aba
for nome_aba in abas_2023:
    nome_base = nome_aba.lower().replace(" ", "_").replace("-", "_").replace("(", "").replace(")", "")
    nome_df = f"df_2023_{nome_base}"
    df_temp = pd.read_excel(caminho_arquivo_2023, sheet_name=nome_aba)

    # Tipagem
    if coluna_categoria in df_temp.columns:
        df_temp[coluna_categoria] = df_temp[coluna_categoria].astype('category')

    for col in colunas_numericas:
        if col in df_temp.columns:
            df_temp[col] = pd.to_numeric(df_temp[col], errors='coerce')

    # Criar coluna Total
    colunas_validas = [col for col in colunas_numericas if col in df_temp.columns]
    df_temp['Total'] = df_temp[colunas_validas].sum(axis=1)

    # Criar coluna Ano
    df_temp['Ano'] = ano_referencia

    # Reorganizar colunas: Nível, Local, Total, Métricas..., Ano
    colunas = list(df_temp.columns)
    geo_col = colunas[1]
    nova_ordem = [coluna_categoria, geo_col, 'Total'] + [c for c in colunas if c not in [coluna_categoria, geo_col, 'Total', 'Ano']] + ['Ano']
    df_temp = df_temp[nova_ordem]

    dfs_2023[nome_df] = df_temp

# Separar por nível e exportar
for nome_df, df in dfs_2023.items():
    if 'Nível' in df.columns:
        for nivel in df['Nível'].dropna().unique():
            nivel_nome = str(nivel).strip().lower().replace(" ", "_")
            nome_final = f"{nome_df}_{nivel_nome}"

            df_nivel = df[df['Nível'] == nivel].copy()

            # Garantir ordenação correta
            colunas = list(df_nivel.columns)
            if 'Total' in colunas:
                colunas.remove('Total')
                colunas.insert(2, 'Total')
            if 'Ano' in colunas:
                colunas.remove('Ano')
                colunas.append('Ano')
            df_nivel = df_nivel[colunas]

            # Exportar
            caminho_saida_final = os.path.join(caminho_saida, f"{nome_final}.xlsx")
            df_nivel.to_excel(caminho_saida_final, index=False)
```

```python
# 📦 Importando bibliotecas
import pandas as pd
import numpy as np

# 📁 Caminhos dos arquivos
dim_geo_path = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/Fato & Dimensao/Dimensão/dim_geografica_expandida.xlsx'
fato_absoluto_path = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/Fato & Dimensao/Fato/fato_absoluto_expandido_corrigido.xlsx'
fato_percentual_path = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/Fato & Dimensao/Fato/fato_percentual_expandido_corrigido.xlsx'
dim_ano_path = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/Fato & Dimensao/Dimensão/dim_ano.xlsx'
output_path = '/content/drive/MyDrive/DNC/Projetos/Becomex/01. Base de dados/SIDRA/TRATADOS/Fato & Dimensao/Refatoradas/'

# 📄 Leitura dos dados
dim_geo = pd.read_excel(dim_geo_path)
fato_abs = pd.read_excel(fato_absoluto_path)
fato_perc = pd.read_excel(fato_percentual_path)
dim_ano = pd.read_excel(dim_ano_path)

# ✅ Padronização dos nomes de coluna
dim_geo.columns = dim_geo.columns.str.strip().str.lower()
dim_ano.columns = dim_ano.columns.str.strip().str.lower()

# 🧱 Tabelas geográficas a partir do 'nível'
def criar_tabela_localizacao(df, nivel_valor, nome_coluna):
    return pd.DataFrame({
        nome_coluna: df[df['nível'] == nivel_valor]['localização'].dropna().drop_duplicates().reset_index(drop=True)
    })

dim_grande_regiao = criar_tabela_localizacao(dim_geo, 'GR', 'grande_regiao')
dim_uf = criar_tabela_localizacao(dim_geo, 'UF', 'uf')
dim_municipio = criar_tabela_localizacao(dim_geo, 'MU', 'municipio')
dim_mesorregiao = criar_tabela_localizacao(dim_geo, 'ME', 'mesorregiao')
dim_microrregiao = criar_tabela_localizacao(dim_geo, 'MI', 'microrregiao')

# 🧱 Dimensão Nível Geográfico
dim_nivel = pd.DataFrame({'nivel': dim_geo['nível'].dropna().unique()})

# 📍 Dimensão Código UF
dim_codigo_uf = pd.DataFrame({'codigo_uf': dim_geo['código uf'].dropna().unique()})

# 🧾 Dimensão Produto
produtos = [
    "Algodão herbáceo (em caroço)", "Arroz (em casca)", "Cacau (em amêndoa)", "Café (em grão) Total",
    "Cana-de-açúcar", "Feijão (em grão)", "Fumo (em folha)", "Laranja", "Milho (em grão)", "Soja (em grão)"
]
dim_produtos = pd.DataFrame({'produto': produtos})

# 📅 Dimensão Ano
dim_ano['ano'] = pd.to_datetime(dim_ano['ano'], format='%Y')
dim_ano.drop_duplicates(inplace=True)

# ⚖️ Dimensão Medidas
def classificar_medida(col):
    col = col.lower()
    if 'tonelada' in col or 'kg' in col:
        return 'toneladas'
    elif 'hectare' in col:
        return 'hectares'
    elif 'valor' in col or 'r$' in col:
        return 'reais'
    elif '%' in col or 'percentual' in col:
        return 'percentual'
    else:
        return 'outro'

def descricao_medida(tipo):
    return {
        'toneladas': 'Peso (tonelada = 1000kg)',
        'hectares': 'Área cultivada (ha)',
        'reais': 'Valor monetário (R$)',
        'percentual': 'Indicador percentual (%)',
        'outro': 'Outra métrica'
    }[tipo]

medidas_numericas = list(set(
    [col for col in fato_abs.columns if fato_abs[col].dtype in ['float64', 'int64']] +
    [col for col in fato_perc.columns if fato_perc[col].dtype in ['float64', 'int64']]
))

dim_medidas = pd.DataFrame({'medida': medidas_numericas})
dim_medidas['tipo'] = dim_medidas['medida'].apply(classificar_medida)
dim_medidas['descricao'] = dim_medidas['tipo'].apply(descricao_medida)
dim_medidas.drop_duplicates(inplace=True)

# 💾 Exportação de todas as dimensões
dim_nivel.to_excel(output_path + "dim_nivel.xlsx", index=False)
dim_codigo_uf.to_excel(output_path + "dim_codigo_uf.xlsx", index=False)
dim_grande_regiao.to_excel(output_path + "dim_grande_regiao_corrigido.xlsx", index=False)
dim_uf.to_excel(output_path + "dim_uf.xlsx", index=False)
dim_municipio.to_excel(output_path + "dim_municipio.xlsx", index=False)
dim_mesorregiao.to_excel(output_path + "dim_mesorregiao.xlsx", index=False)
dim_microrregiao.to_excel(output_path + "dim_microrregiao.xlsx", index=False)
dim_produtos.to_excel(output_path + "dim_produtos.xlsx", index=False)
dim_ano.to_excel(output_path + "dim_ano_formatado.xlsx", index=False)
dim_medidas.to_excel(output_path + "dim_medidas.xlsx", index=False)

print("✅ Todas as dimensões foram geradas e exportadas com sucesso.")
```

```python
def extrair_variavel_unidade(nome_arquivo):
    nome = nome_arquivo.lower()

    if 'colhida' in nome:
        return 'Área colhida', 'hectares'
    elif 'plantada' in nome:
        return 'Área plantada', 'hectares'
    elif 'valor_da_produção' in nome:
        return 'Valor da produção', 'mil reais'
    elif 'rendimento' in nome:
        return 'Rendimento médio', 'kg/hectare'
    elif 'participação' in nome or 'do_total' in nome:
        return 'Participação percentual', '%'
    else:
        return 'Desconhecido', 'Desconhecido'

# Aplica função
fato_agricultura_long[['Variável', 'Unidade']] = fato_agricultura_long['Arquivo Origem'].apply(
    lambda x: pd.Series(extrair_variavel_unidade(x))
)

# Visualiza
display(fato_agricultura_long[['Arquivo Origem', 'Variável', 'Unidade']].drop_duplicates())
```
